```python
from stellar_analysis import plot_cno, plot_phosphorus, plot_potassium, plot_sulfur, plot_nir, plot_ce
starname = 'e53'
observed = 'e53.dat'
rv = -84
```

Hot star, CNO lines are flat

P and S consistent with APSCAP

```python
# Original APSCAP parameters
speclist = ["e53xall.dat"]
label = ["[C/Fe]=-0.41, [N/Fe]=0.43, [O/Fe]=0.34"]

plot_cno(
    synthetic_files=speclist,
    observed_file=observed,
    labels=label,
    starname=starname,
    figsize=(5,6),
    rvoffset = rv,
    yoffset1=0.16, ncorr1=1.2,
    yoffset2=0.142, ncorr2=1.2,
    yoffset3=0.163, ncorr3=1.2,
    ylim1=(0.96,1.01),
    ylim2=(0.88,1.03),
    ylim3=(0.98,1.01)
)
```
